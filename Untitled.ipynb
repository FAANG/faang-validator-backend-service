```python
from typing import Dict, List, Optional, Set
import requests

BIOSAMPLES_SEARCH_URL = (
    "https://www.ebi.ac.uk/biosamples/samples"
    "?size=200&filter=attr%3Aproject%3AFAANG&cursor=*"
)

def fetch_all_samples_batch(etags_cache: Optional[Dict[str, str]] = None) -> List[Dict]:
    """Fetch all FAANG samples via paginated batch request."""
    samples: List[Dict] = []
    url: Optional[str] = BIOSAMPLES_SEARCH_URL
    page = 0

    while url:
        page += 1
        resp = requests.get(url, timeout=120)
        resp.raise_for_status()
        data = resp.json()

        page_samples = data.get("_embedded", {}).get("samples", [])
        for sample in page_samples:
            if etags_cache:
                sample["etag"] = etags_cache.get(sample.get("accession", ""), "")
            samples.append(sample)

        url = data.get("_links", {}).get("next", {}).get("href")

    return samples
```
