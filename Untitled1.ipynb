```python


  import sys
  from collections import defaultdict
  sys.path.insert(0, "/Users/alexey/TREC/be")
  from enrich import enrich_sample

  INDEX = "data_portal_development_4"

  # --- First pass: enrich all samples ---

  resp = es.search(index=INDEX, body={"query": {"match_all": {}}, "size": 500}, scroll="5m")
  scroll_id = resp["_scroll_id"]
  total = resp["hits"]["total"]["value"]
  processed = 0

  while True:
      hits = resp["hits"]["hits"]
      if not hits:
          break

      bulk_body = []
      for hit in hits:
          enriched = enrich_sample(hit["_source"])
          bulk_body.append({"update": {"_index": INDEX, "_id": hit["_id"]}})
          bulk_body.append({"doc": enriched})

      if bulk_body:
          es.bulk(body=bulk_body, refresh=False)

      processed += len(hits)
      if processed % 5000 == 0 or processed == total:
          print(f"  Enriched {processed}/{total}")
      resp = es.scroll(scroll_id=scroll_id, scroll="5m")

  es.indices.refresh(index=INDEX)
  print(f"First pass done: {processed} samples enriched")

  # --- Second pass: resolve derived_sample_ids ---

  print("Resolving derived_sample_ids...")

  # Scroll all derived samples, build parent -> children map
  parent_to_children = defaultdict(list)

  resp = es.search(index=INDEX, body={
      "query": {"exists": {"field": "parent_sample_id"}},
      "_source": ["biosampleId", "parent_sample_id"],
      "size": 500,
  }, scroll="5m")
  scroll_id = resp["_scroll_id"]

  while True:
      hits = resp["hits"]["hits"]
      if not hits:
          break
      for hit in hits:
          parent = hit["_source"].get("parent_sample_id")
          child = hit["_source"].get("biosampleId")
          if parent and child:
              parent_to_children[parent].append(child)
      resp = es.scroll(scroll_id=scroll_id, scroll="5m")

  print(f"  Found {sum(len(v) for v in parent_to_children.values())} derived "
        f"samples across {len(parent_to_children)} parents")

  # Scroll source samples, match with children, build bulk update
  bulk_body = []

  resp = es.search(index=INDEX, body={
      "query": {"term": {"is_source_sample": True}},
      "_source": ["biosampleId"],
      "size": 500,
  }, scroll="5m")
  scroll_id = resp["_scroll_id"]

  while True:
      hits = resp["hits"]["hits"]
      if not hits:
          break
      for hit in hits:
          children = parent_to_children.get(hit["_source"]["biosampleId"])
          if children:
              bulk_body.append({"update": {"_index": INDEX, "_id": hit["_id"]}})
              bulk_body.append({"doc": {"derived_sample_ids": children}})
      resp = es.scroll(scroll_id=scroll_id, scroll="5m")

  # Send bulk updates in chunks
  for i in range(0, len(bulk_body), 2000):
      es.bulk(body=bulk_body[i:i + 2000], refresh=False)
  es.indices.refresh(index=INDEX)
  print(f"  Updated {len(bulk_body) // 2} source samples")

  print("Done!")
```
